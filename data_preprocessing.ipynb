#Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

# Importing the Dataset

```python
dataset = pd.read_csv('sample_data/Data.csv')
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values

print(X)
print(y)
```

# Taking care of missing data

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
imputer.fit(X[:, 1:3])
X[:, 1:3] = imputer.transform(X[:, 1:3])
```

# Encoding categorical data

## Encoding independent variable

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [0])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
```

##Encoding dependent variable

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(y)
```

# Splitting the dataset

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=1)
```

# Feature scaling
### To be applied after data split
Standardization (always between -3 to 3): x<sub>stand</sub> = x - mean(x) / standard deviation (x)

Normalization (always between 0 to 1): x<sub>norm</sub> = x - min(x) / max(x) - min(x)

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(X_train[:,3:])
X_train[:,3:]= scaler.transform(X_train[:,3:])
X_test[:,3:]= scaler.transform(X_test[:,3:])

print(X_train)
print(X_test)
```
